```python
#Data Needed: longitude, latitude, timestamp
#Need to loop 100 times to generate required data points
#Need to append the data in the latitude, longitude, and timestamp lists for it to show up
#Timestamp is in UTC form
```

```python
import requests
```

```python
import time
```

```python
import datetime
```

```python
import matplotlib.pyplot as plt
```

```python
iss_position="https://api.wheretheiss.at/v1/satellites/25544"

latitude_list=[]
longitude_list=[]
timestamp_list=[]

for i in range(100):
    response=requests.get(iss_position)
    data=response.json()

    latitude_value=float(data['latitude'])
    longitude_value=float(data['longitude'])
    timestamp_value=data['timestamp']
    readable_timestamp=datetime.datetime.utcfromtimestamp(timestamp_value)

    latitude_list.append(latitude_value)
    longitude_list.append(longitude_value)
    timestamp_list.append(readable_timestamp)

    print(f"Data point {i+1}: latitude={latitude_value}, longitude={longitude_value}, timestamp={readable_timestamp.strftime('%m-%d-%Y %H:%M:%S UTC')}")

    time.sleep(10)

fig, ax = plt.subplots()

ax.plot(latitude_list, longitude_list, color='green', linestyle='--', marker='o')

ax.set_xlabel('Latitude')
ax.set_ylabel('Longitude')
ax.set_title('Location of the ISS')

plt.show()
```
